---

```txt
bú kè qì 不客气 "don't / no need to be so polite"

bú yòng xiè 不用谢 "don't / no need to thank me"
```

```txt
lǐ        里   in(side); within

lǐ   miàn 里面 the inside area; interior (a location)

wǎng lǐ   往里 toward the inside (a directional motion)
```

```txt
zǒu    走   to walk; to move; to go; to leave

zǒu lù 走路 to walk (as a form of movement or transportation)

sàn bù 散步 to take a walk; to go for a stroll
```
