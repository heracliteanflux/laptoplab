# 2.22 Talk about friends

---

```txt
boyfriend                    nán   péng  yǒu/you    男朋友
country(side)                nóng  cūn              农村
eighteen                     shí   bā               十八
excellent/fantastic/great    tài   bàng  le         太棒了🥹
girlfriend                   nǔü   péng  yǒu/you    女朋友
good friend                  hǎo   péng  yǒu/you    好朋友🧑‍🤝‍🧑
his home/family              tā    jiā              他家
home/family                  jiā   (rén)            家(人)🏠🧑‍🧑‍🧒
interest/hobby               ài    hào              爱好
Mahjong (the game)           má    jiàng            麻将🀄️
my home/family               wǒ    jiā   (rén)      我家(人)🏠🧑‍🧑‍🧒
tv drama                     diàn  shì   jù         电视剧

how <adj>?             duō   <adj>?          多<adj>？
  how old?             duō   dà?             多大？

how many people?       jǐ    kǒu rén?        几口人？
  one  person          yī    kǒu rén         一口人
  five people          wǔ    kǒu rén         五口人
how many animals?      jǐ    zhǐ _?          几只_？

big/old                      dà                     大
different / "not the same"   bù    tóng             不同
lazy                         lǎn                    懒
powerful/severe/serious/awesome/terrible/impressive
                             lì    hai              厉害
small/young                  xiǎo                   小

to dance               tiào  wǔ              跳舞💃
to go to school        shàng xué             上学🎒
to ice skate           huá   bīng            滑冰⛸️
to play Mahjong        dǎ    má   jiàng      打麻将🀄️
to reside/stay/live    zhù   zài             住在
```

```txt
对，我比他小。(duì, wǒ bǐ tā xiǎo) Yes, I'm younger than him.
那是我姐姐，不是我妹妹。(nà shì wǒ jiějie, bú shì wǒ mèimei) That's my older sister, not my younger sister.
你比他小吗？(nǐ bǐ tā xiǎo ma?) Are you younger than him?
你的狗懒妈？(nǐ de gǒu lǎn ma?) Is your dog lazy?
你的好朋友打麻将吗？(nǐ de hǎopéngyou dǎ májiàng ma?) Does your good friend play Mahjong?
你多大？(nǐ duō dà?) How big/old are you?
你家有几口人？(nǐjiā yǒu jǐkǒu rén?) How many people does your family have?
你几点上学？(nǐ jǐdiǎn shàngxué?) What time to you go to school?
你姐姐多大？(nǐ jiějie duō dà?) How big/old is your older sister?
你周末打不打麻将？(nǐ zhōumò dǎ bù dǎ májiàng?) Do you play Mahjong on the weekend?
请问，你家有几口人？(qǐngwèn, nǐjiā yǒu jǐkǒu rén) Excuse me, how many people are in your family?
他多大？(tā duō dà?) How big/old is he?
他会唱歌，不会跳舞。(tā huì chànggē, bú huì tiàowǔ) He knows how to sing, but cannot dance.
他会滑冰吗？(tā huì huábīng ma?) Does he know how to ice skate?
他会滑雪，不会滑冰。(tā huì huáxuě, bú huì huábīng) He knows how to ski, but not ice skate.
他会跳舞。(tā huì tiàowǔ) He knows how to dance.
他今年多大？(tā jīnnián duō dà?) How big/old is he this year?
他妈妈没有车。(tā māma méi yǒu chē) His mom doesn't have a car.
他男朋友很忙。(tā nánpéngyou hěn máng) His boyfriend is busy.
他女儿今年多大？(tā nǔüér jīnnián duō dà?) How big/old is his daughter this year?
他平时跳舞吗？(tā píngshí tiàowǔ ma?) Does he usually dance?
他是我的好朋友。(tā shì wǒ de hǎopéngyou) He is my good friend.
他十八岁。(tā shíbā suì) He's eighteen years old.
他十六岁。(tā shíliù suì) He's sixteen years old.
他特别厉害。(tā tèbié lìhai) He's especially impressive.
她想见他男朋友。(tā xiǎng jiàn tā de nánpéngyou) She wants to meet his boyfriend.
他有什么爱好？(tā yǒu shénme àihào) What are his hobbies?
他在哪里上学？(tā zài nǎlǐ shàngxué?) Where does he go to school?
他在农村长大。(tā zài nóngcūn zhǎngdà) He grew up in the countryside.
他真懒。(tā zhēn lǎn) He's really lazy.
他只有一只猫，没有狗。(tā zhǐ yǒu yīzhǐ māo, méi yǒu gǒu) He only has one cat, no dog.
太棒了，我也喜欢打麻将。(tàibàngle, wǒ yě xǐhuan dǎ májiàng) Great, I also like to play Mahjong.
他们没有女朋友。(tāmen méi yǒu nǔüpéngyou) They don't have girlfriends.
他们住在哪里？(tāmen zhù zài nǎlǐ?) Where do they live?
哇，太棒了。(wā, tàibàngle) Wow, that's great.
王明会滑雪，我会滑冰。(wángmíng huì huáxuě, wǒ huì huábīng) Wang Ming knows how to ski, and I can ice skate.
王明十七岁。(wángmíng shíqī suì) Wang Ming is seventeen years old.
我的狗厉害吗？(wǒ de gǒu lìhai ma?) Is my dog impressive?
我的朋友都在农村。(wǒ de péngyou dōu zài nóngcūn) All my friends are in the countryside.
我的朋友喜欢滑冰。(wǒ de péngyou xǐhuan huábīng) My friend likes to ice skate.
我会弹吉他，他们会唱歌。(wǒ huì tán jítā, tāmen huì chànggē) I know how to play the guitar, and they can sing.
我家有六口人。(wǒjiā yǒu liùkǒu rén) My family has six people.
我家有五口人。(wǒjiā yǒu wǔkǒu rén) My family has five people.
我姐姐的男朋友很聪明。(wǒ jiějie de nánpéngyou hěn cōngming) My older sister's boyfriend is smart.
我没有男朋友。(wǒ méi yǒu nánpéngyou) I don't have a boyfriend.
我男朋友会滑冰，他很厉害。(wǒ nánpéngyou huì huábīng, tā hěn lìhai) My boyfriend knows how to ice skate, he's impressive.
我男朋友有点儿懒。(wǒ nánpéngyou yǒudiǎnr lǎn) My boyfriend's a bit lazy.
我男朋友在纽约上学。(wǒ nánpéngyou zài niǔyuē shàngxué) My boyfriend goes to school in New York.
我上小学，他上中学，我们的学校不同。(wǒ shàng xiǎoxué, tā shàng zhōngxué, wǒmen de xuéxiào bùtóng) I go to elementary school and he goes to middle school, our schools are different.
我想学滑冰。(wǒ xiǎng xué huábīng) I want to learn to ice skate.
我在上学，他在工作。(wǒ zài shàngxué, tā zài gōngzuò) I'm going to school, he's working.
我们的车不同。(wǒmen de chē bùtóng) Our cars are different.
我们的衣服不同。(wǒmen de yīfu bùtóng) Our clothes are different.
我们是学生，在上海上学。(wǒmen shì xuéshēng, zài shànghǎi shàngxué) We're students going to school in Shanghai.
我们最喜欢看电视剧。(wǒmen zuì xǐhuan kàn diànshìjù) We like watching tv dramas the most.
中国的农村特别漂亮。(zhōngguó de nóngcūn tèbié piàoliang) China's countryside is especially beautiful.
```
