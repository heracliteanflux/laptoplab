# 1.1 - Name food and drinks

---

```txt
coffee                                 kā   fēi      咖啡☕️
congee/porridge, rice-based Chinese    zhōu          粥🥣
rice                                   mǐ   fàn      米饭🍚
soup                                   tāng          汤🥣
tea                                    chá           茶🍵
tofu                                   dòu  fǔ/fu    豆腐
water                                  shuǐ          水💦

and; with                              hé            和
this                                   zhè           这

hot                                    rè            热🔥

to be (the Chinese copula)             shì           是
```

---

```txt
米饭和汤。mǐfàn hé tāng Rice and soup.
热茶和咖啡。rè chá hé kāfēi Hot tea and coffee.
水和米饭。shuǐ hé mǐfàn Water and rice.
这是豆腐。zhè shì dòufu This is tofu.
这是热水。zhè shì rè shuǐ This is hot water.
```

---

`shì 是`
* the Chinese copular verb "to be"
* used mainly to link the subject to a noun phrase, not used with adjectives
* used for emphasis or affirmation 我是不知道 wǒ shì bù zhīdào
* used in the shì...de construction to mark emphasis on time, place, manner, etc.

`hé 和`
* normally links noun phrases, not whole clauses
