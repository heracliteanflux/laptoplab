```txt
āiyā, zhèjiàn yīfú/fu yǒudiǎn dà 哎呀，这件衣服有点大。Oh no, these clothes are a bit big.

huānyíng, lǐtiānměi 欢迎，李天美。Welcome, Li Tianmei.
huānyíng, qǐngjìn. 欢迎，请进。Welcome, please come in.
huānyíng wǒmen de xīn tóngshì 欢迎我们的新同事。Welcome our new colleague.

huìyìshì zài èrcéng 会议室在二层。The meeting room is on the second floor.

jīnwǎn lǎobǎn qù bú qù? 今晚老板去不去？Is the boss going tonight?

lǐxiānshēng, nǐhǎo. wángmíng zài bàngōngshì ma? 李先生，你好。王明在办公室吗？Hello Mr. Li. Is Wang Ming in the office?

nǎge bàngōngshì shì wǒde? 哪个办公室是我的？Which office is mine?
nàge huì bù zhòngyào 那个会重要。That meeting/gathering is not important.
nǎge huìyìshì bǐjiào dà? 哪个会议室比较大？Which meeting/gathering room is bigger?
nǎge tóngshì shì xīn tóngshì? 哪个同事是新同事？Which coworker is the new coworker?
nǎshuāng xié bǐjiào shūfú/fu? 哪双鞋比较舒服？Which pair of shoes is more comfortable?

nǐ jīntiān qù xīn gōngsī ma? Are you going to the new company today?
nǐ xiǎng chuān nǎjiàn yīfú/fu? Which clothes do you want to wear?

nǐmen hē bù hē jiǔ? 你们喝不喝酒？Do y'all drink alcohol?

shǎo hē jiǔ 少喝酒。Drink less alcohol.

shéi dài jiǔ? 谁带酒？Who's bringing alcohol?
shéi diǎncài? 谁点菜？Who's ordering?
shéi dìng fànguǎn? 谁订饭馆？Who booked the restaurant?
shéi qǐngle jià? 谁请了假？Who took leave?
shéi qù gàosù jīnglǐ? 谁去告诉经理？Who's going to tell the manager?
shéi shì wángmíng? 谁是王明？Who is Wang Ming?
shéi shì wángxiānshēng? 谁是王先生？Who is Mr. Wang?

shítáng bú zài bā céng 食堂不在八层。The cafeteria is not on the eighth floor.

tā shì gōngsī de fānyì 他是公司的翻译。He's the company's translator.
tā shì nǐ de xīn tóngshì ma? 他是你的新同事吗？Is he your new coworker?

wáng jīnglǐ zài nǎlǐ? 王经理在哪里？Where is manager Wang?

wǒ bù hē jiǔ 我不喝酒。I don't drink alcohol.
wǒ de tóngshì xiànzài hěn máng 我的同事现在很忙。My coworker is busy now.
wǒ fēicháng xǐhuan wǒ de xīn tóngshì 我非常喜欢我的新同事。I really like my new coworker.
wǒ jiào zhàoxiǎoyuè 我叫赵小月。My name is Zhao Xiaoyue.
wǒ jīntiān yào qù xīn gōngsī 我今天要去新公司。I'm going to the new company today.
wǒ xiǎng chuān xīn yīfú/fu 我想穿新衣服。I want to wear new clothes.
wǒ yīnggāi chuān nǎtiáo kùzi? 我应该穿哪条裤子？Which pair of pants should I wear?
wǒ zài shícéng gōngzuò 我在十层工作。I work on the tenth floor.

wǒmen de jīnglǐ tèbié máng 我们的经理特别忙。Our manager is extremely busy.
wǒmen dìng měiguó fànguǎn bā 我们美国饭馆吧。Let's book a table at an American restaurant.
wǒmen xiàwǔ kāihuì ma? 我们下午开会吗？Are we having a meeting this afternoon?

xiān diǎncài 先点菜。Order first.

yīngwén hěn zhòngyào 英文很重要。English is important.

zǎoshànghǎo, wángxiānshēng. 早上好，王先生。Good morning, Mr. Wang.

zhège bàngōngshì shì nǐde 这个办公室是你的。This office is yours.
zhège huì hěn zhòngyào 这个会很重要。This meeting's important.
zhège jìhuà bǐjiào zhòngyào 这个计划比较重要。This plan is rather important.
zhèjiàn chènshān hǎokān ma? Does this shirt look good?
zhè shì fǎyǔ fānyì de bàngōngshì 这是法语翻译的办公室。This is the French translator's office.
```
