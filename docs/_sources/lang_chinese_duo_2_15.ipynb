```txt
āiyā, wǒ de hùzhào zài nǎlǐ? 哎呀，我的护照在哪里？Oh no, where is my passport?

dìtiě dàole 地铁到了。The subway has arrived.
dìtiě dàole ma? 地铁到了吗？Has the subway arrived?
dìtiězhàn zài nàlǐ 地铁站在那里。The subway station is over there.

fēijī jǐdiǎn dào běijīng de jīchǎng? 飞机几点到北京的机场？What time does the plane arrive at Beijing airport?
fēijī jǐdiǎn qǐfēi? 飞机几点起飞？What time does the plane take off?
fēijī qǐfēile ma? 飞机起飞了吗？Has the plane taken off?

gōngjiāochē jǐdiǎn dào? 公交车几点到？What time does the bus arrive?
gōngjiāozhàn zài jiǔdiàn pángbiān 公交站在酒店旁边。The bus stop is next to the hotel.
gōngjiāozhàn zài nàlǐ　公交站在那里。The bus stop is over there.

jiǔdiàn bù yuǎn 酒店不远。The hotel isn't far.
jiǔdiàn pángbiān shì yígè dìtiězhàn 酒店旁边是一个地铁站。There is a subway station next to the hotel.

māma jīchǎng yuǎn ma? 妈妈机场远吗？Mom, is the airport far?

měimei zài xǐ liǎn 妹妹在洗脸。My sister is washing her face.

nǐmen yào zuò chūzūchē ma? 你们要坐出租车吗？Y'all will take a taxi? (Are y'all going to take a taxi?)

qǐng wèn, xiànzài jǐdiǎn? 请问，现在几点？Excuse me, what time is it now?

sāndiǎn shífēn 三点十分。3:10.
sāndiǎn sìshífēn 三点四十分。3:40.

shàngwǔ jiǔdiǎn 上午九点。Nine in the morning.
shàngwǔ jiǔdiǎn èrshífēn 上午九点二十分。9:20 in the morning.
shàngwǔ jiǔdiǎn sìshífēn 上午九点四十分。9:40 in the morning.

shídiǎn wǔshífēn 十点五十分。10:50.

tā shíèrdiǎn èrshífēn dào jiǔdiàn 他十二点二十分到酒店。He arrived at the hotel at 12:20.

wǎnshàng qīdiǎn èrshífēn 晚上七点二十分。7:20 in the evening.
wǎnshàng shíyìdiǎn sìshífēn 晚上十一点四十分。11:40 at night.

wǒ dìdi méi dài hùzhào 我弟弟没带护照。My little brother didn't bring his passport.
wǒ méi dài hùzhào 我没带护照。I didn't bring my passport
wǒ zài nàge dìtiězhàn duìmiàn 我在那个地铁站对面。I'm across from that subway station.

wǒmen bādiǎn èrshífēn chūfā 我们八点二十分出发。We depart at 8:20.
wǒmen de chūzūshē zài nàlǐ 我们的出租车在那里。Our taxi is over there.
wǒmen de fēijī jǐdiǎn qǐfēi? 我们的飞机几点起飞？What time does our plane take off?
wǒmen jǐdiǎn chūfā? 我们几点出发？What time do we depart?
wǒmen jǐdiǎn dào jiǔdiàn? 我们几点到酒店？What time do we arrive at the hotel?
wǒmen liǎng diǎn dào jīchǎng 我们两点到机场。We arrive at the airport at two.
wǒmen shíèr diǎn bàn chīfàn 我们十二点半吃饭。We eat at twelve-thirty.
wǒmen shíèr diǎn chūfā bā 我们十二点出发吧。Let's depart at twelve.
wǒmen shíyī diǎn chūmén 我们十一点出门。We're leaving at eleven.
wǒmen shíyī diǎn qù zuò dìtiě 我们十一点去坐地铁。We're leaving to take the subway at eleven.

xiànzài bā diǎn 现在八点。It's eight now.
xiànzài shíyī diǎn bàn 现在十一点半。It's eleven-thirty now.
xiànzài sìdiǎn wǔshífēn 现在四点五十分。It's now 4:50.
xiànzài wǔ diǎn bàn 现在五点半。It's five-thirty now.
```
