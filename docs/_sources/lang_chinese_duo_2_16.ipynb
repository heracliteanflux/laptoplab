# 2.16 Shop for clothes

---

```txt
belt                                     pí     dài         皮带
black                                    hēi    sè          黑色
brand                                    pái    zi          牌子
car                                      chē                车🚗
different / something else other than    bié    de          别的
dress                                    lián   yī  qún     连衣裙👗
fitting room                             shì    yī  jiān    试衣间
glove                                    shǒu   tào         手套🧤
pair                                     shuāng             双
pants                                    kù     zi          裤子👖
pizza                                    pī     sà          披萨🍕
shoe                                     xié                鞋👟
sock                                     wà     zi          袜子🧦
white                                    bái    sè          白色🤍

    a pair of chopsticks    yì shuāng kuài zi    一双筷子
    a pair of shoes         yì shuāng xié        一双鞋

a little (bit)           yǒu   diǎn r    有点儿
cheap                    pián  yí        便宜
long                     cháng           长
hard                     yìng            硬
thick                    hòu             厚
tight                    jǐn             紧

to be hungry                  è            饿
to be on sale / discounted    dǎ    zhé    打折🈹
to be thirsty                 kě           渴
to drive                      kāi   chē    开车🚗
to shop / go shopping         guàng jiē    逛街
```

---

```txt
hǎoā, wǒmen qù biéde fànguǎn bā 好啊，我们去别的饭馆吧。Sure, let's go to a different restaurant.

hēisè de pídài bú guì 黑色的皮带不贵。The black belt is not expensive.

jīntiān wǒ bù xiǎng kāichē 今天我不想开车。I don't want to drive today.

nàge páizi de yīfú/fu hěn guì 那个牌子的衣服很贵。That brand's clothes are expensive.

nàshuāng shǒutào hěn piányí 那双手套很便宜。That pair of gloves is cheap.
nàshuāng wàzi bùshūfú 那双袜子不舒服。That pair of socks is uncomfortable.

nàtiáo wéijīn tèbié cháng 那条围巾特别长。That scarf is especially long.

nǐ de xié jǐn bù jǐn? 你的鞋紧不紧？Are your shoes tight?
nǐ huí bù huí jiā? 你回不回家？Are you going home?
nǐ kě bù kě? 你渴不渴？Are you thirsty?
nǐ mǎi bù mǎi wàzi? 你买不买袜子？Are you buying socks or not?
nǐ xiǎng guàngjiē ma? 你想逛街吗？Do you want to go shopping?

shìyījiān dà bú dà? 试衣间大不大？Is the fitting room big?

wǒ fēicháng kě, wǒ xiǎng hē shuǐ 我非常渴，我想喝水。I'm extremely thirsty, I want to drink water.
wǒ tèbié è 我特别饿。I'm especially hungry.
wǒ xiǎng mǎi nàtiáo pídài 我想买那条皮带。I want to buy that belt.
wǒ xǐhuan zhètiáo kùzi 我喜欢这条裤子。I like these pants.
wǒ xūyào kùzi, yīfú/fu, hé xié 我需要裤子，衣服，和鞋。I need pants, clothes, and shoes.
wǒ xūyào yìshuāng xīn wàzi 我需要一双新袜子。I need a new pair of socks.
wǒ yào mǎi hēisè de xié 我要买黑色的鞋。I want to buy black shoes.

wǒmen huí jiā bā 我们回家吧。Let's go home.
wǒmen kāichē bā 我们开车吧。Let's drive.
wǒmen mǎi bù mǎi nàshuāng xié? 我们买不买那双鞋？Are we buying that pair of shoes?
wǒmen qù bú qù biéde shāngchǎng? 我们去不去别的商场？Are we going to a different mall?

zhège qiánbāo yǒudiǎnr yìng 这个钱包有点儿硬。This wallet is a bit hard.
zhège yīfú/fu tèbié hòu 这个衣服特别厚。This clothing is especially thick.

zhètiáo xiàngliàn hěn jǐn 这条项链很紧。This necklace is very tight.
zhètiáo liányīqún yǒudiǎnr cháng 这条连衣裙又点儿长。This dress is a bit long.

zhèshuāng shǒutào guì bú guì? 这双手套贵不贵？Is this pair of gloves expensive?
zhèshuāng shǒutào shūfú bù shūfú? 这双手套舒服不舒服？Is this pair of gloves comfortable?
zhèshuāng xié bù dǎ zhé 这双鞋不打折。This pair of shoes is not on sale.
zhèshuāng xié yìng bú yíng? 这双鞋硬不硬？Is this pair of shoes hard?
zhèshuāng xié yǒudiǎnr dà 这双鞋有点儿大。This pair of shoes is a bit big.
```
