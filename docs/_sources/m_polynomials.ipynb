
<b>Comments on nomenclature</b>

In the general theory of polynomials (algebraic geometry), a <b>form</b> of degree $n$ refers to a polynomial homogeneous in that degree. This means that the polynomial only contains degree-$n$ terms. In this context, the term form has a technical meaning. On the other hand, a polynomial equation in degree $n$ may contain lower order terms. We may speak of "the general form of" an equation, but this sense of the term form is not special.

general $n$-th degree polynomial equation in $k$ variables
* first-degree = linear
* second-degree = quadratic
* third-degree = cubic
* fourth-degree = quartic
* fifth-degree = quintic

---
---
---

## Degree-one polynomial equations in several variables

<div class="full-width" style="border: 0.5px solid #0096FF; border-radius: 1rem; padding: 1rem; margin: 1rem 0; box-sizing: border-box;">

<span style="color: #0096FF;"><b>Linear in a single variable</b></span>

The general form of a linear equation in a single variables $x$ is

$Ax + B = 0$ where $A \ne 0$

</div>

$
\begin{aligned}
Ax = -B \iff x = -\frac{B}{A}
\end  {aligned}
$

<div class="full-width" style="border: 0.5px solid #0096FF; border-radius: 1rem; padding: 1rem; margin: 1rem 0; box-sizing: border-box;">

<span style="color: #0096FF;"><b>Linear in two variables</b></span>

The general form of a linear equation in two variables $x, y$ is

$Ax + By + C = 0$ where $A \ne 0 \lor B \ne 0$

</div>

From here we derive the <b>standard form of a line</b> (in which this $C$ is the negative of the $C$ above)

$Ax + By = C$

From here we derive the <b>slope-intercept form of a line</b> $y = mx + b$.

$
\begin{aligned}
Ax + By = C \iff By = -Ax + C \iff y = \underset{m}{\boxed{-\frac{A}{B}}}x + \underset{b}{\boxed{\frac{C}{B}}}
\end  {aligned}
$

From here we derive the <b>point-slope form of a line</b> $y - y_0 = m(x - x_0)$ for a given point $(x_0, y_0)$ and slope $m$.

$y = mx + b \iff b = -mx + y$ for any $(x, y)$. Let's say $x = x_0$ and $y = y_0$. Then $b = -mx_0 + y_0$. Now we can substitute this back into the original equation.

$y = mx + b = mx + (-mx_0 + y_0) = m(x - x_0) + y_0$

<div class="full-width" style="border: 0.5px solid #0096FF; border-radius: 1rem; padding: 1rem; margin: 1rem 0; box-sizing: border-box;">

<span style="color: #0096FF;"><b>Linear in several variables</b></span>

The general form of a linear equation in several variables $x_1, \dotsc, x_n$ is

$a_1 x_1 + a_2 x_2 + \dotsb + a_n x_n + b = 0$ where $a_1 \ne 0 \lor a_2 \ne 0 \lor \dotsb \lor a_n \ne 0$

</div>

From here we derive linear forms in higher dimensions (the plane in three dimensions and the hyperplane in greater-than-three dimensions).

---
---
---

## Degree-two polynomial equations in several variables

<div class="full-width" style="border: 0.5px solid #0096FF; border-radius: 1rem; padding: 1rem; margin: 1rem 0; box-sizing: border-box;">

<span style="color: #0096FF;"><b>General quadratic in one variable</b></span>

The general form of a quadratic equation in a single variable $x$ is

$Ax^2 + Bx + C = 0$ where $A \ne 0$

</div>

<div class="full-width" style="border: 0.5px solid #0096FF; border-radius: 1rem; padding: 1rem; margin: 1rem 0; box-sizing: border-box;">

<span style="color: #0096FF;"><b>General quadratic in two variables</b></span>

The general form of a quadratic equation in two variables $x, y$

with cross terms is

$Ax^2+Bxy+Cy^2+Dx+Ey+F=0$ where $A\ne0\lor B\ne0\lor C\ne0$

without cross terms is

$Ax^2+By^2+Cx+Dy+E=0$ where $A\ne0\lor B\ne0$

</div>

In classical analytic geometry, this is the general equation of a conic section.

---
---
---

## Discriminant

For a degree-$n$ polynomial in a single variable

$p(x) = a_n x^n + \dotsb + a_1 x + a_0$

with roots $r_1, \dotsc, r_n$ the discriminant is defined as

$
\begin{aligned}
\text{Disc}(p) = a_n^{2n - 2} \prod_{i \lt j} (r_i - r_j)^2
\end  {aligned}
$

For the degree-two polynomial equation in a single variable

$p(x) = a x^2 + b x + c$

this gives

$\text{Disc}(p) = b^2 - 4ac$

This tells us about the nature of the roots of the polynomial

$
\begin{aligned}
\text{Disc}(p) &\gt 0 && \text{two distinct real roots} \\
\text{Disc}(p) &=   0 && \text{one repeated real root} \\
\text{Disc}(p) &\lt 0 && \text{two complex roots} \\
\end  {aligned}
$

In the context of degree-$n$ polynomial equations in a single variable, the discriminant is defined as a function of the coefficients. The vanishing of the discriminant is an indicator of degeneracy, and the condition of degeneracy is the multiplicity of the roots.

---
---
---

## Terms

* [ [w](https://en.wikipedia.org/wiki/Abel%E2%80%93Ruffini_theorem) ] Abel-Ruffini Theorem
* [ [w](https://en.wikipedia.org/wiki/Algebraic_equation) ] Algebraic Equation (Polynomial Equation)
* [ [w](https://en.wikipedia.org/wiki/Algebraic_expression) ] Algebraic Expression
* [ [w](https://en.wikipedia.org/wiki/Solution_in_radicals) ] Algebraic Solution
* [ [w](https://en.wikipedia.org/wiki/Bring_radical) ] Bring Radical
* [ [w](https://en.wikipedia.org/wiki/Determinant) ] Determinant
* [ [w](https://en.wikipedia.org/wiki/Discriminant) ] Discriminant
* [ [w](https://en.wikipedia.org/wiki/Algebraic_equation) ] Equation, Algebraic (Polynomial)
* [ [w](https://en.wikipedia.org/wiki/Algebraic_equation) ] Equation, Polynomial (Algebraic)
* [ [w](https://en.wikipedia.org/wiki/Homogeneous_polynomial) ] Form (or Homogeneous Polynomial or Quantic)
* [ [w](https://en.wikipedia.org/wiki/Galois_theory) ] Galois Theory
* [ [w](https://en.wikipedia.org/wiki/Homogeneous_coordinates) ] Homogeneous Coordinates
* [ [w](https://en.wikipedia.org/wiki/Homogeneous_polynomial) ] Homogeneous Polynomial (or Form or Quantic)
* [ [w](https://en.wikipedia.org/wiki/Hypersurface) ] Hypersurface
* [ [w](https://en.wikipedia.org/wiki/Irreducible_polynomial) ] Irreducible Polynomial
* [ [w](https://en.wikipedia.org/wiki/Linear_form) ] Linear Form
* [ [w](https://en.wikipedia.org/wiki/Irreducible_polynomial) ] Polynomial, Irreducible
* [ [w](https://en.wikipedia.org/wiki/Algebraic_equation) ] Polynomial Equation (Algebraic Equation)
* [ [w](https://en.wikipedia.org/wiki/Quadratic_form) ] Quadratic Form
* [ [w](https://en.wikipedia.org/wiki/Quadratic_formula) ] Quadratic Formula
* [ [w](https://en.wikipedia.org/wiki/Quadric) ] Quadric
* [ [w](https://en.wikipedia.org/wiki/Homogeneous_polynomial) ] Quantic (or Form or Homogeneous Polynomial)
* [ [w](https://en.wikipedia.org/wiki/Radical_symbol) ] Radical Symbol
* [ [w](https://en.wikipedia.org/wiki/Resultant) ] Resultant

---
---
---
