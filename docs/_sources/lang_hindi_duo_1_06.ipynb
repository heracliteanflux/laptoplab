# 1.6 - Talk about food

---

```txt
banana               kelā             केला
hello; goodbye       namaste          नमस्ते
no                   nahī̃             नहीं
tea                  cāy              चाय
thank you            dhanyavād        धन्यवाद
water                pānī             पानी
what                 kyā              क्या
who                  kaun             कौन
yes                  hā̃               हाँ
you (intimate)       tū               तू

to drink             pītā/pītī        पीता/पीती
to eat               khātā/khātī      खाता/खाती
to read; to study    paṛhtā/paṛhtī    पढ़ता/पढ़ती
```

---

```txt
cāy aur pānī                                    Tea and water.                             चाय और पानी ।
cāy kaun pītā hai?                              Who drinks tea?                            चाय कौन पीता है ?
cāy kyā hai? pānī kyā hai? maĩ kaun hū̃          What is tea? What is water? Who am I?      चाय क्या है ? पानी क्या है ? मैं कौन हूँ ?
dhanyavād, pīṭar                                Thanks, Peter.                             धन्यवाद, पीटर ।
hā̃, maĩ cāy pītā hū̃                             Yes, I drink tea.                          हाँ, मैं चाय पीता हूँ ।
hā̃, rāj cāy pītā hai                            Yes, Raj drinks tea.                       हाँ, राज चाय पीता है ।
jūliyā khātī kyā hai?                           What does Julia eat.                       जूलिया खाती क्या है ?
jūliyā pānī pītī hai                            Julia drinks water.                        जूलिया पानी पीती है ।
jūliyā seb khātī hai                            Julia eats apple.                          जूलिया सेब खाती है ।
kaun hai?                                       Who's there?                               कौन है ?
kelā aur pānī                                   Banana and water.                          केला और पानी ।
laṛkī kelā khātī hai                            The girl eats the banana.                  लड़की केला खाती है ।
maĩ kelā khātā hū̃                               I eat banana.                              मैं केला खाता हूँ ।
maĩ khātī hū̃                                    I eat.                                     मैं खाती हूँ ।
maĩ pānī pītā hū̃                                I drink water.                             मैं पानी पीता हूँ ।
maĩ paṛhtā hū̃                                   I read.                                    मैं पढ़ता हूँ ।
maĩ paṛhtī hū̃                                   I read.                                    मैं पढ़ती हूँ ।
maĩ seb khātī hū̃                                I eat apple.                               मैं सेब खाती हूँ ।
maĩ ṭhīk hū̃, dhanyavād                          I'm fine, thanks.                          मैं ठीक हूँ, धन्यवाद ।
nahī̃, cāy nahī̃. pānī ṭhīk hai                   No, not tea. Water's fine.                 नहीं, चाय नहीं । पानी ठीक है ।
namaste, maĩ nehā hū̃                            Hi, I'm Neha.                              नमस्ते, मैं नेहा हूँ ।
nehā cāy nahī̃ pītī (hai)                        Neha doesn't drink tea.                    नेहा चाय नहीं पीती (है) ।
nehā cāy pītī hai                               Neha drinks tea.                           नेहा चाय पीती है ।
nehā khātī aur pītī kyā hai?                    What does Neha eat and drink?              नेहा खाती और पीती क्या है ?
nehā paṛhtī hai aur jūliyā khātī hai            Neha reads and Julia eats.                 नेहा पढ़ती है और जूलिया खाती है ।
pānī kaun nahī̃ pītā (hai)?                      Who doesn't drink water?                   पानी कौन नहीं पीता (है) ?
pīṭar pānī pītā hai                             Peter drinks water.                        पीटर पानी पीता है ।
pīṭar paṛhtā hai aur khātā hai                  Peter reads and eats.                      पीटर पढ़ता है और खाता है ।
pīṭar seb khātā hai                             Peter eats apple.                          पीटर सेब खाता है ।
rāj kelā khātā hai                              Raj eats banana.                           राज केला खाता है ।
rāj pānī pītā hai                               Raj drinks water.                          राज पानी पीता है ।
rāj vaha kitāb nahī̃ paṛhtā hai                  Raj doesn't read that book.                राज वह किताब नहीं पढ़ता है ।
ṭhīk hai, namaste                               Okay, bye.                                 ठीक है, नमस्ते ।
tū khātā hai                                    You eat.                                   तू खाता है ।
tū pānī pītī hai                                You drink water.                           तू पानी पीती है ।
tū pīṭar hai                                    You're Peter.                              तू पीटर है ।
vaha kaun hai?                                  Who's that?                                वह कौन है ?
vaha kelā khātā hai                             He eats banana.                            वह केला खाता है ।
vaha kyā hai?                                   What's that?                               वह क्या है ?
vaha laṛkā pānī pītā hai                        That boy drinks water.                     वह लड़का पानी पीता है ।
vaha laṛkī cāy pītī hai                         That girl drinks tea.                      वह लड़की चाय पीती है ।
vaha laṛkī kitāb paṛhtī hai                     That girl reads a book.                    वह लड़की किताब पढ़ती है ।
vaha laṛkī kaun hai?                            Who's that girl?                           वह लड़की कौन है ?
yaha ādmī pānī pītā hai aur kitāb paṛhtā hai    This man drinks water and reads a book.    यह आदमी पानी पीता है और किताब पढ़ता है ।
yaha ek kelā hai                                This is a banana.                          यह एक केला है ।
yaha laṛkā paṛhtā hai                           This boy reads.                            यह लड़का पढ़ता है ।
```
