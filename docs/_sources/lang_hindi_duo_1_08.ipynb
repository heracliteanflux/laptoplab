# 1.8 - Introduce people

---

```txt
America             amerikā          अमेरिका
brother             bhāī             भाई
daughter            beṭī             बेटी
day                 din              दिन
every               har              हर
from                se               से
his/hers            uskā/uskī        उसका/उसकी
home                ghar             घर
in                  mẽ               में
India               bhārat           भारत
my/mine             merā/merī        मेरा/मेरी
name                nām              नाम
school              skūl             स्कूल
sister              bahan            बहन
son                 beṭā             बेटा
your/yours          terā/terī        तेरा/तेरी

(possessive)        kā/kī            का/की

to go (to)          jātā/jātī        जाता/जाती
to live; to stay    rahatā/rahatī    रहता/रहती
```

---

```txt
bhārat merā ghar hai                 India is my home.                भारत मेरा घर है ।
hā̃, maĩ bhārat mẽ rahatā hū̃          Yes, I live in India.            हाँ, मैं भारत में रहता हूँ ।
har ādmī skūl jātā hai               Every man goes to school.        हर आदमी स्कूल जाता है ।
jūliyā kā ghar                       Julia's house.                   जूलिया का घर ।
maĩ amerikā mẽ rahatā hū̃             I live in America.               मैं अमेरिका में रहता हूँ ।
maĩ amerikā mẽ rahatī hū̃             I live in America.               मैं अमेरिका में रहती हूँ ।
merā bhāī terā bhāī nahī̃ hai         My brother is not your brother.  मेरा भाई तेरा भाई नहीं है ।
merā nām jūliyā hai                  My name's Julia.                 मेरा नाम जूलिया है ।
merī bahan kī kitāb                  My sister's book.                मेरी बहन की किताब ।
merī beṭī skūl jātī hai              My daughter goes to school.      मेरी बेटी स्कूल जाती है ।
nehā rāj kī beṭī hai                 Neha is Raj's daughter.          नेहा राज की बेटी है ।
nehā terī bahan hai                  Neha is your sister.             नेहा तेरी बहन है ।
pīṭar bhārat se nahī̃ hai             Peter isn't from India.          पीटर भारत से नहीं है ।
rāj kī beṭī                          Raj's daughter.                  राज की बेटी ।
terā aur merā                        Yours and mine.                  तेरा और मेरा ।
terā nām kyā hai?                    What's your name?                तेरा नाम क्या है ?
terī bahan seb khātī hai             Your sister eats apples.         तेरी बहन सेब खाती है ।
uskā beṭā har din skūl jātā hai      Her son goes to school every day. उसका बेटा हर दिन स्कूल जाता है ।
uskā bhāī āmir hai                   His brother is Amir.             उसका भाई आमिर है ।
uskā bhāī bhārat mẽ rahatā hai       His brother lives in India.      उसका भाई भारत में रहता है ।
uskī bahan amerikā mẽ rahatī hai     His sister lives in America.     उसकी बहन अमेरिका में रहती है ।
uskī beṭī ghar mẽ hai                Her daughter is in the house.    उसकी बेटी घर में है ।
vaha amerikā jātī hai                She goes to America.             वह अमेरिका जाती है ।
vaha bhārat mẽ hai                   He is in India.                  वह भारत में है ।
vaha bhārat mẽ rahatā hai            He lives in India.               वह भारत में रहता है ।
vaha laṛkā bhārat mẽ nahī̃ rahatā hai That boy doesn't live in India.  वह लड़का भारत में नहीं रहता है ।
vaha laṛkī skūl se ghar jātī hai     That girl goes home from school. वह लड़की स्कूल से घर जाती है ।
vaha merā beṭā hai                   He's my son.                     वह मेरा बेटा है ।
vaha merī beṭī hai aur yaha terī beṭī hai That is my daughter and this is your daughter. वह मेरी बेटी है और यह तेरी बेटी है ।
vaha uskā skūl hai                   That's his school.               वह उसका स्कूल है ।
vaha uskī kitāb hai                  That's his book.                 वह उसकी किताब है ।
yaha merā ghar hai                   This is my home.                 यह मेरा घर है ।
yaha merā ghar nahī̃ hai              This isn't my home.              यह मेरा घर नहीं है ।
yaha merī bahan hai                  This is my sister.               यह मेरी बहन है ।
yaha merī beṭī hai                   This is my daughter.             यह मेरी बेटी है ।
yaha merī kitāb hai                  This is my book.                 यह मेरी किताब है ।
yaha nehā kā beṭā hai                This is Neha's son.              यह नेहा का बेटा है ।
yaha terī kitāb hai                  This is your book.               यह तेरी किताब है ।
yaha uskā din hai                    This is his day.                 यह उसका दिन है ।
```
