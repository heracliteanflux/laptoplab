# 1.5 - Use possessive pronouns

---

```txt
also; too                yě                   也
dad/father               bà    bà/ba          爸爸👨      
daughter                 nǔü   ér             女儿👧
husband                  lǎo   gōng           老公
middle school student    zhōng xué   shēng    中学生
mom/mother               mā    mā/ma          妈妈👩
son                      ér    zi             儿子👦
wife                     lǎo   pó             老婆

(possessive particle)    de                   的
```

---

```txt
你是英国人，对不对？nǐ shì yīngguórén, duìbúduì? You're British, right?
他的儿子也是学生。tā de érzi yě shì xuéshēng His son is also a student.
她的老公是中国人。tā de lǎogōng shì zhōngguórén Her husband is Chinese.
我的爸爸是医生。wǒ de bàba shì yīshēng My dad's a doctor.
这是我的女儿。zhè shì wǒ de nǔüér This is my daughter.
```
