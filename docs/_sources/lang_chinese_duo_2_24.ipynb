```txt
弟弟，你快洗脸。(dìdi, nǐ kuài xǐliǎn) Brother, hurry up and wash your face.
今天你有作业吗？(jīntiān nǐ yǒu zuòyè ma?) Do you have homework today?
今天是星期一。(jīntiān shì xīngqīyī) Today is Monday.
今天我的午饭是面条。(jīntiān wǒ de wǔfàn shì miàntiáo) Today my lunch consists of noodles.
老婆，我先工作，然后给你打电话。(lǎopó, wǒ xiān gōngzuò, ránhòu gěi nǐ dǎ diànhuà) Wife, first I'll work and then I'll give you a call.
妈妈现在很困。(māma xiànzài hěn kùn) Mother is sleepy now.
没问题，我去洗碗。(méiwèntí, wǒ qù xǐ wǎn) No problem, I'll go wash the dishes.
你别点外卖。(nǐ bié diǎn wàimài) Don't order takeout.
你别喝汽水。(nǐ bié hē qìshuǐ) Don't drink soda.
你吃一个鸡蛋。(nǐ chī yígè jīdàn) You eat an egg.
你给我煮一个鸡蛋。(nǐ gěi wǒ zhǔ yígè jīdiàn) Boil an egg for me.
你快吃午饭。(nǐ kuài chī wǔfàn) You eat lunch quickly.
你快穿衣服。(nǐ kuài chuān yīfu) Hurry up and put on your clothes.
你困吗？(nǐ kùn ma?) Are you sleepy?
你去帮张经理。(nǐ qù bāng zhāng jīnglǐ) You go help manager Zhang.
你先帮他，然后帮我。(nǐ xiān bāng tā, ránhòu bāng wǒ) First help him, then help me.
你先换衣服。(nǐ xiān huàn yīfu) First change your clothes.
你想放一些辣椒吗？(nǐ xiǎng fàng yìxiē làjiāo ma?) Do you want to add some chili peppers?
你现在有时间吗？(nǐ xiànzài yǒu shíjiān ma?) Do you have time now?
你早点儿睡觉。(nǐ zǎodiǎnr shuìjiào) Go to bed earlier.
谁想帮他？(shéi xiǎng bāng tā?) Who wants to help him?
他去接爷爷和奶奶。(tā qù jiē yéye hé nǎinai) He's going to pick up grandpa and grandma.
他洗了澡，然后换了衣服。(tā xǐlezǎo, ránhòu huànle yīfu) He bathed and then got changed.
外卖很便宜。(wàimài hěn piányì) Takeout is cheap.
我别喝汽水。(wǒ bù hē qìshuǐ) I don't drink soda.
我不喜欢吃鸡蛋。(wǒ bù xǐhuan chī jīdàn) I don't like to eat eggs.
我今天非常忙，没有时间。(wǒ jīntiān fēicháng máng, méi yǒu shíjiān) I'm extremely busy today, I don't have time.
我马上出门。(wǒ mǎshàng chūmén) I'm heading out now.
我马上到家。(wǒ mǎshàng dào jiā) I'm getting home any moment now.
我马上下班。(wǒ mǎshàng xiàbān) I'm leaving work soon.
我每天都散步。(wǒ méitiān dōu sànbù) I take a walk each and every day.
我去接你吧。(wǒ qù jiē nǐ bā) I'll go pick you up.
我去买汽水，今晚我们开派对吧。(wǒ qù mǎi qìshuǐ, jīnwǎn wǒmen kāi pàiduì bā) I'm going to buy soda, let's have a party tonight.
我太困了。(wǒ tài kùn le) I'm too sleepy.
我先回家，然后点外卖。(wǒ xiān huí jiā, ránhòu diǎn wàimài) I'll first come home, then order takeout.
我想吃两个鸡蛋。(wǒ xiǎng chī liǎngge jīdàn) I want to eat two eggs.
我星期三要上班。(wǒ xīngqīsān yào shàngbān) I'm going to work on Wednesday.
我应该换一件绿色的衣服。(wǒ yīnggāi huàn yíjiàn lùüsè de yīfu) I should change into a green outfit.
我应该先接儿子，然后回家。(wǒ yīnggāi xiān jiē érzi, ránhòu huí jiā) I should pick up my son first, then return home.
我应该在哪里接他？(wǒ yīnggāi zài nǎlǐ jiē tā?) Where should I pick him up?
我们点外卖吧。(wǒmen diǎn wàimài bā) Let's order takeout.
我们点外卖，怎么样？(wǒmen diǎn wàimài, zěnmeyàng?) How about we order takeout?
我们马上吃饭。(wǒmen mǎshàng chīfàn) We are about to eat.
这些辣椒不好吃。(zhèxiē làjiāo bù hǎochī) These chili peppers aren't good.
```
