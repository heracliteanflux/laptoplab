# Standard Chinese (Mandarin) Phonotactics

---

30 phonemes
* 19 consonants
* 3 glides/semi-vowels
* 1 syllabic consonant
* 2 vowels
* 5 tonemes

or 36 phonemes
* 23 consonants
* 3 glides
* 5 vowels
* 5 tonemes

5 tones/tonemes
* flat
* rising
* falling-rising
* falling
* neutral/toneless - the toneless phoneme is rare, but toneless syllables do occur often in spoken Chinese as allotones (tonal allophones) of syllables which would normally have one of the four tones

In principle, this many phonemes could produce tens of thousands of syllables. But because Standard Chinese phonotactics is so restrictive, there are far fewer.

```python
30**3 # 27,000
36**3 # 46,656
```

Although Japanese has a very small inventory of syllables, a single morpheme is usually composed of many syllables (agglutination); and so Japanese can create many words. On the other hand, a Chinese morpheme usually consists of a single syllable, so 1200 syllables corresponds to only 1200 phonetically distinct morphemes. How does Chinese deal with this? Written Chinese doesn't have a problem, because written Chinese uses morphemic characters (logograms) instead of phonetic ones and so there is no ambiguity in meaning. Spoken Chinese relies on (1) context, (2) set phrases, (3) noun suffixes, (4) complementary morphemes.

## `zhī 之 汁 知 蜘`

```txt
之
meaning: to set off; to go; to head toward a place
top:    止 zhǐ "to stop (modern); foot (oracle bone script)"
bottom: 一 yī  "one (modern); ground, starting point (oracle bone script)"



huàn yán zhī    in other words    换言之

    tā hěn jié jiǎn. huàn yán zhī, tā bù xǐhuan làng fèi qián
    he's frugal. in other words, he doesn't like to waste money
    他很节俭。换言之，他不喜欢浪费钱

X,Y zhī lèi    such as (X and Y) / things of that type (like X and Y)    X,Y之类

    wǒ xǐhuan píngguǒ, xiāngjiāo zhīlèi de shuǐguǒ
    I like fruits such as apples and bananas
    我喜欢苹果，香蕉，之类的水果

zhī suǒ yǐ    之所以    the reason why; the reason for that; why that happens

    tā zhī suǒ yǐ chéng gōng, shì yīn wèi tā cóng bù fàng qì
    The reason he's successful is because he never gives up.
    他之所以成功，是因为他从不放弃

X zhī wài    outside (of) / besides / apart from X     X之外

    gōng zī zhī wài hái yǒu jiǎng jīn    beside salary there's also a bonus    工资之外还有奖金

X zhī yī     one of the group of X                     X之一

    fāng fǎ   zhī yī    one of the methods       方法之一
    yōu  diǎn zhī yī    one of the advantages    优点之一
    yuán yīn  zhī yī    one of the reasons       原因之一

X zhī zhōng    in/among/within X    X之中

    wǒmen zhī zhōng tā zuì nǔ lì    He's the hardest-working among us    我们之中他最努力

zǒng zhī    in short; all in all; to sum up    总之

    wǒmen tǎolùnle hěnduō. zǒng zhī, xiàngmù kěxíng
    we discussed many things. in conclusion, the project is feasible
    我们讨论了很多。总之，项目可行。



bù    liǎo liǎo zhī    不了了之         "not finished, then 之"; to let something end without resolving it; to brush it off
chéng gōng zhī  mǔ                     the mother of success (is failure)                  成功之母
jūn   zǐ   bù   zhī tā xiāng           a gentleman doesn't go to strange lands             君子不之他乡
shì   yǐ   zhì  cǐ  wú kě    zhī yě    things have come to this, there is nowhere to go    事已至此，无可之也
yī    bì   zhī  lì                     "the strength of an arm"; a helping hand            一臂之力
```
