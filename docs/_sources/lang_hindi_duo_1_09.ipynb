# 1.9 - Describe your family

---

```txt
child/kid                  baccī                       बच्ची
children/kids              bacce                       बच्चे
daughters                  beṭiyā̃                      बेटियाँ
father                     pitā                        पिता
friend                     dost                        दोस्त
grandfather                dādā                        दादा
grandfother                nānā                        नाना
grandmother                dādī                        दादी
grandmother                nānī                        नानी
mother                     mā̃                          माँ
our(s)                     hamārā / hamārī / hamāre    हमारा हमारी हमारे
sisters                    bahanẽ                      बहनें
sons                       beṭe                        बेटे
their; his/her (honorific) unka/unki/unke              उनका/उनकी/उनके
together                   sāth mẽ                     साथ में
you                        āp                          आप
your                       āpkā/āpkī/āpke              आपका/आपकी/आपके
your                       tumhārā/tumhārī/tumhāre     तुम्हारा/तुम्हारी/तुम्हारे

(possessive plural)        ke                          के

to live; to stay (plural)    rahate    रहते
```

```txt
āp unkī mā̃ haĩ                       You're their mom.                   आप उनकी माँ हैं ।
āp mere pitā haĩ                     You're my dad.                      आप मेरे पिता हैं ।
āpke pitā aur āpkī mā̃                Your dad and mom.                   आपके पिता और आपकी माँ ।
āpkī bahan merī dost hai             Your sister is my friend.           आपकी बहन मेरी दोस्त है ।
āpkī mā̃ bhārat mẽ rahatī haĩ         Your mom lives in India.            आपकी माँ भारत में रहती हैं ।
bacce seb khāte haĩ                  Children eat apples.                बाच्चे सेब खाते हैं ।
bacce skūl jāte hai                  Children go to school.              बच्चे स्कूल जाते हैं ।
jūliyā aur nehā bahanẽ haĩ           Julia and Neha are sisters.         जूलिया और नेहा बहनें हैं ।
jūliyā ek baccĩ hai                  Julia's a kid.                      जूलिया एक बच्ची है ।
jūliyā ke bhāī                       Julia's brothers.                   जूलिया के भाई ।
kitābẽ merī dost haĩ                 Books are my friends.               किताबें मेरी दोस्त हैं ।
maĩ tumhārā nānā hū̃                  I'm your grandpa.                   मैं तुम्हारी नाना हूँ ।
maĩ tumhārī bahan hū̃                 I'm your sister.                    मैं तुम्हारी बहन हूँ ।
maĩ unkā beṭā hū̃                     I'm his/her/their son.              मैं उनका बेटा हूँ ।
mere bhāī kitābẽ paṛhte haĩ          My brothers read books.             मेरे भाई किताबें पढ़ते हैं ।
mere dādā-dādī bhārat mẽ rahate haĩ  My grandparents live in India.      मेरे दादा-दादी भारत में रहते हैं ।
mere do beṭe haĩ                     I have two sons.                    मेरे दो बेटे हैं ।
mere pitā bhārat mẽ rahate haĩ       My dad lives in India.              मेरे पिता भारत में रहते हैं ।
mere tīn bhāī haĩ                    I have three brothers.              मेरे तीन भाई हैं ।
merī bahanẽ skūl jātī haĩ            My sisters go to school.            मेरी बहनें स्कूल जाती हैं ।
merī beṭiyā̃ ghar jātī haĩ            My daughters go home.               मेरी बेटियाँ घर जाती हैं ।
merī do beṭiyā̃ haĩ                   I have two daughters.               मेरी दो बेटियाँ हैं ।
nānī pānī pītī haĩ                   Grandma drinks water.               नानी पानी पीती हैं ।
rāj aur nehā amerikā nahī̃ rahate haĩ Raj and Neha don't live in America. राज और नेहा अमेरिका में नहीं रहते हैं ।
rāj ke beṭe                          Raj's sons.                         राज के बेटे ।
terī beṭiyā̃ amerikā mẽ rahatī haĩ    Your daughters live in America.     तेरी बेटियाँ अमेरिका मेँ रहती हैं ।
terī do bahanẽ haĩ                   You have two sisters.               तेरी दो बहनें हैं ।
tumhārā nām kyā hai?                 What's your name?                   तुम्हारा नाम क्या है ?
tumhārā skūl bhārat mẽ hai           Your school is in India.            तुम्हारा स्कूल भारत में है ।
unke seb                             His/her/their apples.               उनके सेब ।
unkī mā̃ aur unke pitā                His/her/their mom and dad.          उनकी माँ और उनके पिता ।
uskī mā̃ kī tīn beṭiyā̃ haĩ            His/her mom has three daughters.    उसकी माँ की तीन बेटियाँ हैं ।
uskī tīn beṭiyā̃ haĩ                  He/she has three daughters.         उसकी तीन बेटियाँ हैं ।
vaha hamārī bahan hai                She's our sister.                   वह हमारी बहन है ।
vaha tumhārī kitāb hai               That's your book.                   वह तुम्हारी किताब है ।
ve mere nānā haĩ                     He's my grandpa.                    वे मेरे नाना हैं ।
ve merī dādī haĩ                     She's my grandma.                   वे मेरी दादी हैं ।
ve merī nānī haĩ                     She's my grandma.                   वे मेरी नानी हैं ।
yaha tumhārā kelā hai                This is your banana.                यह तुम्हारा केला है ।
ye mere dādā haĩ                     This is my grandpa.                 ये मेरे दादा हैं ।
ye merī mā̃ haĩ                       This is my mom.                     ये मेरी माँ हैं ।
```
