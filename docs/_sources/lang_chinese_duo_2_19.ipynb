# 2.19 Expressing possession

---

```txt
backpack                    shū   bāo                书包🎒
blue                        lán   sè                 蓝色
dictionary                  cí    diǎn               词典
green                       lùü   sè                 绿色
identification/ID card      shēn  fèn       zhèng    身份证🆔
just (now)                  gāng  gāng/gang          刚刚
Miss                        xiǎo  jiě/jie            小姐
name                        míng  zi                 名字
red                         hóng  sè                 红色
umbrella                    yǔ    sǎn                雨伞🌂
water bottle                shuǐ  bēi                水杯
Zhang Hua                   zhāng huá                张华

    several books           jǐ  běn shū         几本书
    three dictionaries      sān běn cí  diǎn    三本词典

to be (called) / go by      jiào                     叫
to know/understand          zhī   dào                知道
to lose                     diū                      丢
to realize/discover/find    fā    xiàn               发现
to return/repay             huán                     还
to seem (to be) / to be likely    hǎo  xiàng         好像
```

---

```txt
嗨，张华。(hāi, zhāng huá) Hi, Zhang Hua.
明天我要去还电脑。(míngtiān wǒ yào qù huán diànnǎo) I intend to go return the computer tomorrow.
那是李老师的。(nà shì lǐ lǎoshī de) That's Teacher Li's.
那是谁的书？(nà shì shéi de shū?) Whose book is that?
那是张华吗？(nà shì zhānghuá ma?) Is that Zhang Hua?
那些衣服是谁的？(nàxiē yīfú shì shéi de?) Whose clothes are those?
你不知道吗？(nǐ bù zhīdào ma?) You don't know?
你的狗叫什么名字？(nǐ de gǒu jiào shénme míngzi?) What is your dog's name?
你的书好像在图书馆。(nǐ de shū hǎoxiàng zài túshūguǎn) Your book seems to be in the library.
你的英文书是绿色的吗？(nǐ de yīngwén shū shì lùüsè de ma?) Is your English book green?

你的雨伞是红色吗？(nǐ de yǔsǎn shì hóngsè ma?) Is your umbrella red?

你丢了什么？(nǐ diūle shénme?) What did you lose?
你发现了什么？(nǐ fāxiànle shénme?) What did you realize?
你刚刚在哪里？(nǐ gānggang zài nǎlǐ?) Where were you just now?
你叫什么名字？(nǐ jiào shénme míngzi?) What is your name?
请问，我的身份证在这里吗？(qǐngwèn, wǒ de shēnfènzhèng zài zhèlǐ ma?) Excuse me, is my ID card here?
谁的电脑是黑色的？(shéi de diànnǎo shì hēisè de?) Whose computer is black?
谁的帽子是红色的？(shéi de màozi shì hóngsè de?) Whose hat is red?

谁的雨伞是白色？(shéi de yǔsǎn shì báisè?) Whose umbrella is white?

谁是张华？(shéi shì zhānghuá?) Who is Zhang Hua?
手机和台灯都是李老师的。(shǒujī hé táidēng dōu shì lǐ lǎoshī de) Both the phone and lamp are Teacher Li's.
她的帽子好像在宿舍。(tā de màozi hǎoxiàng zài sùshè) Her hat seems to be in the dorm.
她的书包是蓝色的吗？(tā de shūbāo shì lánsè de ma?) Is her backpack blue?
她的雨伞特别可爱。(tā de yǔsǎn tèbié kěài) Her umbrella is especially cute.
他刚刚在宿舍。(tā gānggang zài sùshè) He was just in the dorm.
他还了三本书。(tā huánle sānběn shū) He returned three books.
他叫王明。(tā jiào wángmíng) He's called Wang Ming.
她也不知道。(tā yě bù zhīdào) She doesn't know either.
我不知道。🤷 (wǒ bù zhīdào) I don't know.

我的词典不是绿色。(wǒ de cídiǎn bú shì lùüsè) My dictionary isn't green.

我的衣服是蓝色的。(wǒ de yīfú shì lánsè de) My clothes are blue.
我可以借你词典吗？(wǒ kěyǐ jiè nǐ de cídiǎn ma?) Can I borrow your dictionary?
我没丢手机。(wǒ méi diū shǒujī) I didn't lose my phone.
我要去还书。(wǒ yào qù huán shū) I intend to go return the book.
我要去还书，这些书是图书馆的。(wǒ yào qù huán shū, zhèxiē shū shì túshūguǎn de) I intend to go return the books, they are the library's.
我在那里发现了一个帽子。(wǒ zài nàlǐ fāxiànle yígè màozi) I found a hat there.
小姐，你叫什么名字？(xiǎojiě, nǐ jiào shénme míngzi?) Miss, what is your name?
赵小月也不知道。(zhàoxiǎoyuè yě bù zhīdào) Zhao Xiaoyue also doesn't know.
这个电脑是谁的？(zhège diànnǎo shì shéi de?) Whose computer is this?
这个护照是谁的？(zhège hùzhào shì shéi de?) Whose passport is this?

这个钱包好像是赵小月的。(zhège qiánbāo hǎoxiàng shì zhàoxiǎoyuè de) This wallet seems to be Zhao Xiaoyue's.

这个手机不是张华的。(zhège shǒujī bú shì zhānghuá de) This isn't Zhang Hua's phone.
这个书包好像是你的。(zhège shūbāo hǎoxiàng shì nǐ de) This backpack seems to be yours.
这个书包很重。(zhège shūbāo hěn zhòng) This backpack is heavy.
这是谁的？(zhè shì shéi de?) Whose is this?
这是谁的雨伞？(zhè shì shéi de yǔsǎn?) Whose umbrella is this?

这些东西好像王明的。(zhèxiē dōngxī hǎoxiàng wángmíng de) These things seem to be Wang Ming's.
```
