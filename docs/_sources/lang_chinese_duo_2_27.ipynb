# 2.27 - Use indirect objects

---

```txt
email                           yóu   jiàn               邮件📧
essay                           zuò   wén                作文📓
everyone                        měi   ge  rén            每个人
group                           xiǎo  zǔ                 小组
Italian language (spoken)       yì    dà  lì  yǔ         意大利语🇮🇹
Italian person                  yì    dà  lì  rén        意大利人🇮🇹
Italy                           yì    dà  lì             意大利🇮🇹
last night                      zuó   wǎn                昨晚
pen                             bǐ                       笔🖊️
problem/question                wèn   tí                 问题🙋
question/topic/subject/title    tí    mù                 题目
thought/idea                    xiǎng fǎ                 想法💡
when                            shén  me  shí hòu/hou    什么时候

an email    yì fēng yóu jiàn    一封邮件
a pen       yì zhī bǐ           一支笔

see you tomorrow                míng tiān jiàn    明天见
what to do?                     zěn  me   bàn?    怎么办？

to ask                          wèn                      问
to choose                       xuǎn                     选
to complete/finish              wán   chéng              完成✅
to discuss, to talk about       tǎo   lùn                讨论
to give                         gěi                      给
to lend, to borrow              jiè                      借
to translate                    fān   yì                 翻译
to turn in, to submit           jiāo                     交
```
