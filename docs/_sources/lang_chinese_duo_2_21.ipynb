# 2.21 - Talk about the weather

---

```txt
beach/coast/seaside      hǎi  biān               海边🏝️
coat                     dà   yī                 大衣🧥
cafe / coffee shop       kā   fēi  guǎn          咖啡馆☕️
door                     mén                     门🚪
forecast                 yù   bào                预报
how (is/are)?            zěn  me   yàng?         怎么样？
pretty/quite             tǐng                    挺
raincoat                 yǔ   yī                 雨衣
sweater                  máo  yī                 毛衣
this winter              jīn  nián dōng  tiān    今年冬天
this year                jīn  nián               今年
too (much)               tài (...  le)           太(...了)
weather forecast         tiān qì   yù    bào     天气预报☀️

sunny                    qíng tiān               晴天☀️
warm                     nuǎn huó/huo            暖和
windy                    yǒu  fēng               有风

to close                 guān                    关
to rain                  yǒu  yǔ                 有雨🌧️
to remember              jì   dé/de              记得
to wear (accessories)    dài                     戴
```

---

```txt
北京的冬天挺冷的。(běijīng de dōngtiān tǐng lěng de) Beijing's winter is pretty cold.
别跟他一起去。(bié gēn tā yìqǐ qù) Don't go with him.
别关门。(bié guān mén) Don't close the door.
海边太冷了。(hǎibiān tài lěng le) The beach is too cold.
记得穿大衣。(jìde chuān dàyī) Remember to wear your coat.
记得穿厚裤子。(jìde chuān hòu kùzi) Remember to wear thick pants.
记得关门。(jìde guān mén) Remember to close the door.
记得看天气预报。(jìde kàn tiānqìyùbào) Remember to watch the weather forecast.
今年春天太热。(jīnnián chūntiān tài rè) This spring is too hot.
今年春天有点儿冷。(jīnnián chūntiān yǒudiǎnr lěng) This spring is a bit cold.
今年的天气不好(的)。(jīnnián de tiānqì bù hǎo (de)) This year's weather isn't good.
今年的天气挺好(的)。(jīnnián de tiānqì tǐng hǎo (de)) This year's weather is pretty good.
今年的天气真好(的)。(jīnnián de tiānqì zhēn hǎo (de)) This year's weather is really good.
今年冬天很冷。(jīnnián dōngtiān hěn lěng) This winter is cold.
今年夏天太热了。(jīnnián xiàtiān tài rè le) This summer is too hot.
今天的天气挺好的。下个星期的天气挺热的。(jīntiān de tiānqì tǐng hǎo de. xiàge xīngqī de tiānqì tǐng rè de) Today's weather is pretty good. Next week's weather is pretty hot.
今天你戴不戴帽子？(jīntiān nǐ dài bú dài màozi?) Are you wearing a hat today?
今天上午有风。(jīntiān shàngwǔ yǒufēng) It's windy this morning.
今天是晴天。我不需要雨伞。(jīntiān shì qíngtiān. wǒ bù xūyào yǔsǎn) It's sunny today. I don't need an umbrella.
今天太冷了。(jīntiān tài lěng le) It's too cold today.
今天晚上有雨。记得带雨伞。(jīntiān wǎnshàng yǒuyǔ. jìde dài yǔsǎn) It's raining tonight. Remember to bring your umbrella.
今天晚上有雨吗？(jīntiān wǎnshàng yǒuyǔ ma?) Is it going to rain this evening?
今天下午有风。(jīntiān xiàwǔ yǒufēng) It is windy this afternoon.
今天香港挺冷的。星期六纽约挺热的。(jīntiān xiānggǎng tǐng lěng de. xīngqīliù niǔyuē tǐng rè de) Today Hong Kong is pretty cold. Saturday New York is pretty hot.
今天有雨吗？(jīntiān yǒuyǔ ma?) Is it raining today?
老家的天气怎么样？(lǎojiā de tiānqì zěnmeyàng?) How is the weather at home?
门在哪里？(mén zài nǎlǐ?) Where's the door?
明天(的)天气挺好的。(míngtiān (de) tiānqì tǐng hǎode) Tomorrow's weather will be pretty good.
明天很暖和。(míngtiān hěn nuǎnhuó) Tomorrow is warm.
明天下午有雨。(míngtiān xiàwǔ yǒuyǔ) It will rain tomorrow afternoon.
明天也是晴天。(míngtiān yě shì qíngtiān) Tomorrow is also sunny.
明天有风吗？(míngtiān yǒufēng ma?) Is tomorrow going to be windy?
哪个咖啡馆比较安静？(nǎge kāfēiguǎn bǐjiào ānjìng?) Which cafe is quieter?
你看了天气预报吗？(nǐ kànle tiānqìyùbào ma?) Did you watch the weather forecast?
纽约的天气怎么样？(niǔyuē de tiānqì zěnmeyàng?) How is New York's weather?
上海的春天比较暖和。(shànghǎi de chūntiān bǐjiào nuǎnhuo) Spring in Shanghai is rather warm.
上海的夏天挺热的。(shànghǎi de xiàtiān tǐng rè (de)) Summer in Shanghai is pretty hot.
她的毛衣太旧。(tā de máoyī tài jiù) Her sweater is too old.
天气挺冷(的)。(tiānqì tǐng lěng (de)) The weather is quite cold.
天气挺热(的)。(tiānqì tǐng rè (de)) The weather is quite hot.
我不想去海边。(wǒ bù xiǎng qù hǎibiān) I don't want to go to the beach.
我不常看天气预报。(wǒ bùcháng kàn tiānqìyùbào) I don't often watch the weather forecast.
我常常去那个咖啡馆。(wǒ chángcháng qù nàge kāfēiguǎn) I often go to that cafe.
我穿了毛衣。(wǒ chuānle máoyī) I wore a sweater.
我可以关门吗？(wǒ kěyǐ guān mén ma?) Can I close the door?
我想戴这条蓝色的围巾。(wǒ xiǎng dài zhètiáo lánsè de wéijīn) I want to wear this blue scarf.
我喜欢跟妈妈一起去游泳。(wǒ xǐhuan gēn māma yìqǐ qù yóuyǒng) I like to go swimming with mom.
我需要一件雨衣。(wǒ xūyào yíjiàn yǔyī) I need a raincoat.
我应该穿那件绿色毛衣。(wǒ yīnggāi chuān nàjiàn lùüsè máoyī) I should wear that green sweater.
我在网上看天气预报。(wǒ zài wǎngshàng kàn tiānqìyùbào) I'm watching the weather forecast online.
我们关不关窗户？(wǒmen guān bù guān chuānghù) Do we close the window?
我们先看天气预报吧。(wǒmen xiān kàn tiānqìyùbào bā) Let's first watch the weather forecast.
星期六我们去海边吧。(xīngqīliù wǒmen qù hǎibiān bā) Let's go to the beach Saturday.
星期六有雨，对不对？(xīngqīliù yǒuyǔ, duìbúduì?) It will rain on Saturday, right?
这个假期挺热的。(zhège jiàqī tǐng rè de) This holiday is pretty hot.
这个秋天很热。(zhège qiūtiān hěn rè) This autumn is hot.
这个周末是晴天吗？(zhège zhōumò shì qíngtiān ma?) Is it going to be sunny this weekend?
这个周末有风。(zhège zhōumò yǒufēng) This weekend is going to be windy.
这件大衣挺厚的。(zhèjiàn dàyī tǐng hòu de) This coat is pretty thick.
昨天(的)天气怎么样？(zuótiān (de) tiānqì zěnmeyàng?) How was the weather yesterday?
```
