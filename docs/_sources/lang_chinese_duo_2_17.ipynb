# 2.17 Attend a concert

---

```txt
abroad/international / out of the country    guó   wài            国外
before (in time) /previously/ago             yǐ    qián           以前
city                                         chéng shì            城市🌆
concert                                      yǎn   chàng huì      演唱会
entrance                                     rù    kǒu            入口
holiday/vacation                             jià   qī             假期
online                                       zài   wǎng  shàng    在网上
plan                                         jì    huà            计划
seat                                         zuò   wèi            座位💺
souvenir                                     jì    niàn  pǐn      纪念品
ticket                                       piào                 票🎫
ticket booth/office                          shòu  piào  chù      售票处
which/what                                   nǎ    xiē            哪些

difficult                                    nán                  难☹️
excited                                      xīng  fèn            兴奋🥰

to begin/start                               kāi   shǐ            开始
to be held/happening (at)                    jǔ    xíng           举行
to meet (up (with))                          jiàn  miàn           见面
to park                                      tíng  chē            停车🅿️
to ski                                       huá   xuě            滑雪🎿
to take photos                               pāi   zhào  piàn/pian     拍照片
to wait (for)                                děng                 等
to wait in line / queue                      pái   duì            排队
```

```txt
bàngqiú bǐjiào nán 棒球比较难。Baseball is rather difficult.

běijīng hé niǔyuē dōu shì dà chéngshì 北京和纽约都是大城市。Beijing and New York are both big cities.

bié zàiwǎngshàng mǎi piào 别在网上买票。Don't buy tickets online.

děng wǒ 等我。Wait for me.

duì, huódòng zài tǐyùguǎn jǔxíng 对，活动在体育馆举行。Yes, the event is being held at the gymnasium.

huódòng zài nǎlǐ jǔxíng? 活动在哪里举行？Where is the event being held?

lánqiú bù nán 篮球不难。Basketball isn't difficult.

míngtiān wǒ hé péngyǒu yào qù guówài 明天我和朋友要去国外。My friend and I are going abroad tomorrow.

nǎxiē yùndòng bǐjiào nán? 哪些运动比较难？Which sports are relatively difficult?
nàzhǒng yùndòng hěn guì 那种运动很贵。That kind of sport is expensive.

nǐ bǐ wǒ xīngfèn 你比我兴奋。You're more excited than me.
nǐ huì pāi zhàopiàn ma? 你会拍照片吗？Can you take photos?
nǐ xiǎng qù nǎxiē bówùguǎn? 你想去哪些博物馆？Which museum do you want to go to?
nǐ xiǎng zài nǎlǐ tíngchē? 你想在哪里停车？Where do you want to park?
nǐ xǐhuan nǎxiē chéngshì? 你喜欢哪些城市？Which cities do you like?
nǐ xīngfèn ma? 你兴奋吗？Are you excited?
nǐ yīnggāi zài nàlǐ tíngchē 你应该在那里停车。You should park there.

rénmen tèbié xīngfèn 人们特别兴奋。People are especially excited.

rùkǒu hěn dà 入口很大。The entrance is large.
rùkǒu yuǎn ma? 入口远吗？Is the entrance far?

tā zài gōngjiāozhàn děng tā de péngyǒu 她在公交站等她的朋友。She's at the bus stop waiting for her friend.
tā xiǎng xué huáxuě 他想学滑雪。He wants to learn to ski.
tā yào jiàn nǎxiē jiārén? 他要见哪些家人？Which family members is he going to meet?

wǒ bú huì huáxuě 我不会滑雪。I can't ski.
wǒ de zuòwèi zài nàlǐ 我的座位在那里。My seat is there.
wǒ de zuòwèi zài nǐ de zuòwèi qiánmiàn 我的座位在你的座位前面。My seat is in front of your seat.
wǒ lǎogōng tèbié xǐhuan huáxuě 我老公特别喜欢滑雪。My husband especially likes skiing.
wǒ tèbié xīngfèn 我特别兴奋。I'm especially excited.
wǒ xiǎng gēn gēshǒu pāizhàopiàn 我想跟歌手拍照片。I want to take a photo with the singer.
wǒ zài tíngchēchǎng děng nǐmen 我在停车场等你们。I'm waiting for y'all at the parking lot.

wǒmen jǐdiǎn jiànmiàn? 我们几点见面？What time shall we meet up?
wǒmen sìdiǎn jiànmiàn bā 我们四点见面吧。Let's meet up at four.
wǒmen yìqǐ mǎi piào bā 我们一起买票吧。Let's buy tickets together.
wǒmen zài gōngyuán pángbiān tíngchē bā 我们在公园旁边停车吧。Let's park next to the park.
wǒmen zài nǎlǐ mǎi piào 我们在哪里买票？Where do we buy tickets?
wǒmen zài nǎlǐ páiduì? 我们在哪里排队？Where do we wait?
wǒmen zài nǎlǐ tíngchē? 我们在哪里停车？Where do we park?
wǒmen zài tíngchēchǎng páiduì 我们在停车场排队。We are waiting in the parking lot.
wǒmen zàiwǎngshàng kàn yǎnchànghuì bā 我们在网上看演唱会吧。Let's watch the concert online.

yǎnchànghuì bādiǎn kāishǐ 演唱会八点开始。The concert starts at eight.
yǎnchànghuì zài gōngyuán jǔxíng 演唱会在公园举行。The concert is being held at the park.

zhè shì nǐ de zuòwèi 这是你的座位。This is your seat.

zhège jiàqī nǐ xiǎng zuò shénme? 这个假期你想做什么？What do you want to do this holiday?

zhèzhǒng yùndòng nán bù nán? 这种运动难不难？Is this kind of sport difficult.
```
