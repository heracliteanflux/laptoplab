# 1.7 - Form the plural

---

```txt
2             do        दो (२)
3             tīn       तीन (३)

bananas       kele      केले
books         kitābẽ    किताबें
boys          laṛke     लड़के
girls         laṛkiyā̃   लड़कियाँ
these         ye        ये
those         ve        वे
we            ham       हम
women         auratẽ    औरतें
you           tum       तुम

(you)      are    ho        हो
(we, they) are    haĩ       हैं

to drink (plural)             pīte      पीते
to eat (plural)               khāte     खाते
to read; to study (plural)    paṛhte    पढ़ते
```

---

```txt
ādmī tīn kele khāte haĩ            The men eat three bananas.                 आदमी तीन केले खाते हैं ।
auratẽ kitābẽ paṛhtī haĩ           The women read books.                      औरतें किताबें पढ़ती हैं ।
do ādmī kele khāte haĩ             Two men eat bananas.                       दो आदमी केले खाते हैं ।
do auratẽ                          Two women.                                 दो औरतें ।
do kele                            Two bananas.                               दो केले ।
ham cāy pīte haĩ aur seb khāte haĩ We drink tea and eat apples.               हम चाय पीते हैं और सेब खाते हैं ।
ham kele aur seb khāte haĩ         We eat bananas and apples.                 हम केले और सेब खाते हैं ।
ham laṛke haĩ                      We are boys.                               हम लड़के हैं ।
ham ṭhīk nahī̃ haĩ                  We aren't okay.                            हम ठीक नहीं हैं ।
kitābẽ kaun paṛhtā hai?            Who reads the books?                       किताबें कौन पढ़ता है ?
kyā laṛkiyā̃ aur laṛke paṛhte haĩ?  What do the girls and boys read?           क्या लड़कियाँ और लड़के पढ़ते हैं ?
kyā ye tīn seb haĩ?                Are these three apples?                    क्या ये तीन सेब हैं ?
laṛke cāy nahī̃ pīte (haĩ)          The boys don't drink tea.                  लड़के चाय नहीं पीते (हैं) ।
laṛkiyā̃ paṛhtī haĩ                 The girls read.                            लड़कियाँ पढ़ती हैं ।
laṛkiyā̃ kyā khātī haĩ?             What do the girls eat?                     लड़कियाँ क्या खाती हैं ?
rāj aur nehā pānī pīte haĩ         Raj and Neha drink water.                  राज और नेहा पानी पीते हैं ।
tīn seb                            Three apples.                              तीन सेब ।
tum cāy pīte ho                    Y'all drink tea.                           तुम चाय पीते हो ।
tum khāte ho                       You eat.                                   तुम खाते हो ।
tum kyā pīte ho?                   What do you drink?                         तुम क्या पीते हो ?
tum pānī pītī ho                   You drink water.                           तुम पानी पीती हो ।
tum paṛhte ho                      Y'all read.                                तुम पढ़ते हो ।
ve auratẽ haĩ?                     They are women.                            वे औरतें है ।
ve auratẽ kaun haĩ?                Who are those women?                       वे औरतें कौन है ?
ve kaun haĩ?                       Who are they?                              वे कौन है ?
ve kitābẽ paṛhtī haĩ               They read books.                           वे किताबें पढ़ती हैं ।
ve laṛke kele khāte haĩ            Those boys eat bananas.                    वे लड़के केले खाते हैं ।
ve vaha kitāb paṛhte haĩ           They read that book.                       वे वह किताब पढ़ते हैं ।
ye seb haĩ                         These are apples.                          ये सेब हैं ।
ye tīn ādmī haĩ aur ve do ādmī haĩ These are three men and those are two men. ये तीन आदमी हैं और वे दो आदमी हैं ।
ye laṛke kitābẽ paṛhte haĩ         These boys read books.                     ये लड़के किताबें पढ़ते हैं ।
ye laṛkiyā̃ pānī nahī̃ pītī haĩ      These girls don't drink water.             ये लड़कियाँ पानी नहीं पीती हैं ।
```
