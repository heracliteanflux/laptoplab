# 2.20 - Express feelings

---

```txt
black/red tea                          hóng chá            红茶☕️
Chinese character                      hàn  zì             汉字
chore/housework                        jiā  wù             家务
fast food                              kuài cān            快餐🍱
fruit salad                            shuǐ guǒ  shā lā    水果沙拉
interesting/fun/meaningful             yǒu  yì   si        有意思
Japanese language (spoken)             rì   yǔ             日语🇯🇵
novel                                  xiǎo shuō           小说
oh my!                                 tiān nà!            天呐‼️
salad                                  shā  lā             沙拉🥗
Sichuan                                sì   chuān          四川
Sichuan person (to be from Sichuan)    sì   chuān rén      四川人
snake                                  shé                 蛇🐍
spider                                 zhī  zhū            蜘蛛🕷️
the most / -est                        zuì                 最
usually / in one's spare time          píng shí            平时

dangerous                              wēi  xiǎn           危险⚠️
easy                                   róng yì             容易
boring                                 wú   liáo           无聊🥱

to chat                                liáo tiān           聊天💬
to fear / be afraid (of/to)            pà                  怕😟
to hate                                tǎo  yàn            讨厌👎
to write                               xiě                 写✍️
```

---

```txt
钓鱼真无聊。(diàoyú zhēn wúliáo) Fishing is really boring.
汉字最好看。(hànzì zuì hǎokàn) Chinese characters are the most good-looking.
红茶最好喝。(hóngchá zuì hǎohē) Black tea is the tastiest.
开车特别难。(kāichē tèbié nán) Driving is difficult.
那种语言最容易。(nǎzhǒng yǔyán zuì róngyì) Which language is the easiest.
你常常看小说吗？(nǐ chángcháng kàn xiǎoshuō ma?) Do you frequently read novels?
你常常做家务吗？(nǐ chángcháng zuò jiāwù ma?) Do you frequently do chores?
你平时应该多看电影。(nǐ píngshí yīnggāi duō kàn diànyìng) You should watch more movies in your spare time.
你是四川人吗？(nǐ shì sìchuānrén ma?) Are you from Sichuan?
你应该少聊天。(nǐ yīnggāi shǎo liáotiān) You should chat less.
你在看什么小说？(nǐ zài kàn shénme xiǎoshuō?) What novel are you reading?
你在写什么？(nǐ zài xiě shénme?) What are you writing?
你最讨厌做什么？(nǐ zuì tǎoyàn zuò shénme) What do you hate doing the most?
跑步特别无聊。(pǎobù tèbié wúliáo) Running is especially boring.
平时我不打乒乓球。(píngshí wǒ bù dǎ pīngpāngqiú) I don't usually play ping pong.
日语最难。(rìyǔ zuì nán) Japanese is the most difficult.
沙拉最健康。(shālā zuì jiànkāng) Salad is the healthiest.
说西班牙语比较容易。(shuō xībānyáyǔ bǐjiào róngyì) Speaking Spanish is rather easy.
四川菜最辣。(sìchuān cài zuì là) Sichuan food is the spiciest.
他不喜欢聊天。(tā bù xǐhuan liáotiān) He doesn't like to chat.
他平时喜欢运动吗？(tā píngshí xǐhuan yùndòng ma?) Does he usually like to exercise?
他特别喜欢快餐，快餐很好吃。(tā tèbié xǐhuan kuàicān, kuàicān hěn hǎochī) He especially likes fast food, fast food is tasty.
天呐，那是大蜘蛛。(tiānnà, nà shì dà zhīzhū) Oh my, that is a big spider.
王老师，学汉字真难。(wáng lǎoshī, xué hànzì zhēn nán) Teacher Wang, learning Chinese characters is really difficult.
我不讨厌洗碗。(wǒ bù tǎoyàn xǐ wǎn) I don't hate washing dishes.
我不做家务。(wǒ bù zuò jiāwù) I don't do chores.
我哥哥不怕蛇。(wǒ gēge bù pà shé) My brother isn't afraid of snakes.
我老婆最讨厌沙拉。(wǒ lǎopó zuì tǎoyàn shālā) My wife hates salad the most.
我怕吃药。(wǒ pà chī yào) I'm afraid to take medicine.
我讨厌运动。(wǒ tǎoyàn yùndòng) I hate exercise.
我最怕兔子。(wǒ zuì pà tùzi) I fear rabbits the most.
我最讨厌唱歌。(wǒ zuì tǎoyàn chànggē) I hate singing the most.
我最讨厌蛇。(wǒ zuì tǎoyàn shé) I hate snakes the most.
我最讨厌洗衣服。(wǒ zuì tǎoyàn xǐ yīfú) I hate washing clothes the most.
我最讨厌做家务。(wǒ zuì tǎoyàn zuò jiāwù) I hate doing chores the most.
我们不常聊天。(wǒmen bùcháng liáotiān) We don't chat often.
我们常常聊天。(wǒmen chángcháng liáotiān) We frequently chat.
写汉字最有趣。(xiě hànzì zuì yǒuqù) Writing Chinese characters is the most interesting.
星期六我和家人要做家务。(xīngqīliù wǒ hé jiārén yào zuò jiāwù) My family and I will do chores on Saturday.
学语言不容易。(xué yǔyán bù róngyì) Learning a language is not easy.
学中文真容易。(xué zhōngwén zhēn róngyì) Learning Chinese is really easy.
养狗危险吗？(yǎng gǒu wēixiǎn ma?) Is raising a dog dangerous?
这本小说真有趣。(zhèběn xiǎoshuō zhēn yǒuqù) This novel is really interesting.
蜘蛛危险吗？(zhīzhū wēixiǎn ma?) Are spiders dangerous?
中文比日语难。(zhōngwén bǐ rìyǔ nán) Chinese is more difficult than Japanese.
周末你做家务吗？(zhōumò nǐ zuò jiāwù ma?) Do you do chores on the weekend?
做沙拉不难。(zuò shālā bù nán) Making salad isn't difficult.
```
