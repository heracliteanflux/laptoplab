# 1.5 - Form basic sentences

---

```txt
I        maĩ      मैं

am       hū̃       हूँ
is       hai      है

a/an     ek       एक
and      aur      और
apple    seb      सेब
book     kitāb    किताब
boy      laṛkā    लड़का
girl     laṛkī    लड़की
man      ādmī     आदमी
okay     ṭhīk     ठीक
one      ek       एक
that     vaha     वह
this     yaha     यह
woman    aurat    औरत
```

---

```txt
ek ādmī aur ek aurat         A man and a woman.       एक आदमी और एक औरत ।
ek kitāb aur ek seb          A book and an apple.     एक किताब और एक सेब ।
maĩ ek laṛkā hū̃              I'm a boy.               मैं एक लड़का हूँ ।
maĩ jūliyā hū̃                I'm Julia.               मैं जूलिया हूँ ।
maĩ laṛkī hū̃                 I'm a girl.              मैं लड़की हूँ ।
maĩ ṭhīk hū̃                  I'm fine.                मैं ठीक हूँ ।
nehā ek aurat hai            Neha is a woman.         नेहा एक औरत है ।
nehā ṭhīk hai                Neha is fine.            नेहा ठीक है ।
rāj aur nehā                 Raj and Neha.            राज और नेहा ।
rāj ek ādmī hai              Raj is a man.            राज एक आदमी है ।
rāj ek laṛkā hai             Raj is a boy.            राज एक लड़का है ।
seb aur kitāb                Apple and book.          सेब और किताब ।
vaha ādmī aur yaha aurat     That man and this woman. वह आदमी और यह औरत ।
vaha ek seb hai              That's an apple.         वह एक सेब है ।
vaha laṛkā ṭhīk hai          That boy is fine.        वह लड़का ठीक है ।
vaha laṛkī aur yaha laṛkī    That girl and this girl. वह लड़की और यह लड़की ।
vaha pīṭar hai               That's Peter.            वह पीटर है ।
vaha ṭhīk hai                That's fine.             वह ठीक है ।
yaha ādmī                    This man.                यह आदमी ।
yaha aurat ṭhīk hai          This woman is fine.      यह औरत ठीक है ।
yaha ek laṛkā hai            This is a boy.           यह एक लड़का है ।
yaha ek laṛkī hai            This is a girl.          यह एक लड़की है ।
yaha ek seb hai              This is an apple.        यह एक सेब है ।
yaha jūliyā hai              This is Julia.           यह जूलिया है ।
yaha laṛkī aur vaha laṛkā    This girl and that boy.  यह लड़की और वह लड़का ।
```
