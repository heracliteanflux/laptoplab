# 1.4 - Discuss your courses

---

```txt
class                kè                  课
college student      dà    xué  shēng    大学生
he                   tā                  他
math                 shù   xué           数学
music                yīn   yuè           音乐🎵
she                  tā                  她

to attend (class)    shàng               上
to like              xǐ    huān          喜欢😍
```

```txt
对，我上英文课。duì, wǒ shàng yīngwén kè Yes, I attend English class.
对，我是大学生。duì, wǒ shì dàxuéshēng Yes, I'm a college student.
你是大学生，对不对？nǐ shì dàxuéshēng, duìbúduì? You're a college student, right?
你喜欢数学课，对不对？nǐ xǐhuān shùxué kè, duìbúduì? You like math class, right?
我不上数学课。wǒ bú shàng shùxué kè I don't attend math class.
我喜欢音乐课。wǒ xǐhuān yīnyuè kè I like music class.
我喜欢中文课，你呢？wǒ xǐhuān zhōngwén kè, nǐ ne? I like Chinese class, how about you?
```
