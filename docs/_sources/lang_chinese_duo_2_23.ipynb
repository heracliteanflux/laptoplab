# 2.23 Describe your daily routine

```txt
author/writer                       zuò   jiā                  作家
bike/bicycle                        zì    xíng   chē           自行车🚴
car key                             chē   yào    shi           车钥匙🚗🔑
comic                               màn   huà                  漫画
comic book                          màn   huà    shū           漫画书
daily / every day                   měi   tiān                 每天
diary                               rì    jì                   日记📓
every afternoon                     měi   tiān   xià  wǔ       每天下午
every night/evening                 měi   tiān   wǎn  shàng    每天晚上
good night                          wǎn   ān                   晚安😴
key                                 yào   shi                  钥匙🔑
news                                xīn   wén                  新闻📰

convenient/accessible               fāng  biàn                 方便

sometimes                           yǒu   shí    hòu/hou       有时候
while / at the same time            yì    biān                 一边

to bathe                            xǐ    zǎo                  洗澡🛀
to brush one's teeth                shuā  yá                   刷牙🪥
to cycle / ride a bike/bicycle      qí    zì     xíng chē      骑自行车🚴
to forget (something)               wàng                       忘
to get up                           qǐ    chuáng               起床
to get off work / finish work       xià   bān                  下班
to go to bed/sleep                  qù    shuì   jiào          去睡觉
to go to work / start work          shàng bān                  上班💼
to walk                             zǒu   lù                   走路🚶
```

```txt
出门的时候，记得带钥匙。Remember to bring the key when you go out.
带钥匙。(dài yàoshi) Bring the key.
多走路。(duō zǒulù) Walk more.
记得刷牙。(jìde shuāyá) Remember to brush your teeth.
开车特别方便。(kāichē tèbié fāngbiàn) Driving is especially convenient.
开车特别快也比较方便。(kāichē tèbié kuài yě bǐjiào fāngbiàn) Driving is especially fast and also rather convenient.
开车也很方便。(kāichē yě hěn fāngbiàn) Driving is also convenient.
老婆，晚安。(lǎopó, wǎnān) Goodnight, wife.
你几点下班？(nǐ jǐdiǎn xiàbān?) What time do you get off work?
你几点洗澡？(nǐ jǐdiǎn xǐzǎo?) What time do you bathe?
你平时几点睡觉？(nǐ píngshí jǐdiǎn shuìjiào?) What time do you usually go to bed?
你忘了什么？(nǐ wàngle shénme?) What did you forget?
你忘了水杯吗？(nǐ wàngle shuǐbēi ma?) Did you forget the water bottle?
你晚上十点洗澡吗？(nǐ wǎnshàng shídiǎn xǐzǎo ma?) Do you bathe at 10:00 in the evening?
你早上看新闻吗？(nǐ zǎoshàng kàn xīnwén ma?) Do you watch the news in the morning?
女儿，晚安。(nǔüér, wǎnān) Goodnight, daughter.
平时你写日记吗？(píngshí nǐ xiě rìjì ma?) Do you usually write in a diary?
起床。(qǐchuáng) Get up.
起床太难了。(qǐchuáng tài nán le) Getting up is too difficult.
去睡觉。(qù shuìjiào) Go to bed.
刷牙非常重要。(shuāyá fēicháng zhòngyào) Brushing one's teeth is extremely important.
他不想去睡觉，他在给男朋友打电话。(tā bù xiǎng qùshuìjiào, tā zài gẽi nánpéngyou dǎ diànhuà) He doesn't want to go to bed, he's calling his boyfriend.
他平时八点半起床。(tā píngshí bādiǎnbàn qǐchuáng) He usually gets up at 8:30.
王明不在办公室，他下班了。(wángmíng bú zài bàngōngshì, tā xiàbānle) Wang Ming isn't in his office, he finished work.
我不会骑自行车。(wǒ bú huì qí zìxíngchē) I don't know how to ride a bike.
我很饿，他也很饿。(wǒ hěn è, tā yě hěn è) I'm hungry and he's hungry too.
我没带钥匙。(wǒ méi dài yàoshi) I didn't bring the key.
我每天都运动。(wǒ měitiān dōu yùndòng) I exercise each and every day.
我每天晚上都写日记。(wǒ měitiānwǎnshàng xiě rìjì) I write in my diary each and every night.
我没有自行车。(wǒ méi yǒu zìxíngchē) I don't have a bicycle.
我七点上班。(wǒ qīdiǎn shàngbān) I go to work at 7:00.
我去游泳，他也去游泳。(wǒ qù yóuyǒng, tā yě qù yóuyǒng) I go swimming an he goes swimming too.
我忘了钥匙，也忘了钱包。(wǒ wàngle yàoshi, yě wàngle qiánbāo) I forgot my key and wallet.
我喜欢这个作家，我老婆也喜欢他。(wǒ xǐhuan zhège zuòjiā, wǒ lǎopó yě xǐhuan tā) I like this author, and so does my wife.
我有时候一边运动，一边听歌。(wǒ yǒushíhou yìbiān yùndòng, yìbiān tīng gē) Sometimes I exercise while listening to music.
我有时候做晚饭。(wǒ yǒushíhòu zuò wǎnfàn) I make dinner sometimes.
我在看这个作家的小说。(wǒ zài kàn zhège zuòjiā de xiǎoshuō) I'm reading this author's novel.
我早上七点起床。(wǒ zǎoshàng qīdiǎn qǐchuáng) I get up at 7:00 in the morning.
我们每天都刷牙。(wǒmen měitiān dōu shuāyá) We brush our teeth each and every day.
我们一边做饭，一边唱歌。(wǒmen yìbiān zuòfàn, yìbiān chànggē) We cook while singing.
星期六我不上班。(xīngqīliù wǒ bù shàngbān) I don't go to work on Saturday.
洗澡很舒服。(xǐzǎo hěn shūfu) Bathing feels good.
赵小月，你几点起床？(zhàoxiǎoyuè, nǐ jǐdiǎn qǐchuáng?) Zhao Xiaoyue, what time do you get up?
走路很健康，也很舒服。(zǒulù hěn jiànkāng, yě hěn shūfu) Walking is healthy and it feels good too.
坐出租车特别方便。(zuò chūzūchē tèbié fāngbiàn) Taking a taxi is especially convenient.
```
